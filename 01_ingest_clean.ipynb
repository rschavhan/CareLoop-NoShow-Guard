```python
from typing_extensions import Reader
import csv

with open("/content/Appointment csv.csv", newline="") as files:
  reader = csv.DictReader(files)
  rows = list(reader)


print("Columns: ",reader.fieldnames)
print("Total rows:", len(rows))
print("First record:", rows[0])
```

```python
def parse_row(row):
  row = row.copy()

  row["age"] = int(row["age"])
  row["fee_inr"] = int(row["fee_inr"])
  row["prior_no_shows"] = int(row["prior_no_shows"])
  row["distance_km"] = float(row["distance_km"])
  return row



print(parse_row(rows[0]))
```

```python
accepted = []
rejected = []

for row in rows:
  try:
    parsed_row = parse_row(row)
    accepted.append(parsed_row)
  except ValueError:
    rejected.append(row)

print(f"Successfully parsed: {len(accepted)} rows")
print(f"Rejected (due to missing/invalid values): {len(rejected)} rows")
```

```python
import json

with open("accepted.json", "w") as file:
  json.dump(accepted, file)

with open("rejected.json", "w") as file:
  json.dump(rejected, file,indent=2)
```

```python
print("original rows", len(rows))
print("accepted rows", len(accepted))
print("rejected rows", len(rejected))

print("first rejected record:")
print(json.dumps(rejected[0], indent=2))
```
